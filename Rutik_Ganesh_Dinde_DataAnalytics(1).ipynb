```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display plots inside the notebook
%matplotlib inline

print("Libraries imported successfully.")
```

## 1. Load Dataset

Place your CSV file in the same folder as this notebook and change `DATA_FILE` if required.

Example:
`data.csv`


```python
# Change this filename to your actual dataset filename
DATA_FILE = "data.csv"

try:
    df = pd.read_csv(DATA_FILE)
    print("Dataset loaded successfully.")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
except FileNotFoundError:
    print(f"File not found: {DATA_FILE}")
    print("Please place the CSV file in the same folder as this notebook.")
```

## 2. Basic Dataset Information

```python
# View first five records
if 'df' in globals():
    display(df.head())

# Dataset dimensions
if 'df' in globals():
    print("Dataset shape:", df.shape)

# Column names
if 'df' in globals():
    print("\nColumn names:")
    print(df.columns.tolist())

# Data types
if 'df' in globals():
    print("\nData types:")
    print(df.dtypes)
```

## 3. Data Quality Check

```python
if 'df' in globals():
    print("Missing values:")
    display(df.isnull().sum().sort_values(ascending=False))

    print("\nDuplicate rows:", df.duplicated().sum())

    print("\nStatistical summary:")
    display(df.describe(include='all').T)
```

## 4. Data Cleaning

```python
if 'df' in globals():
    # Remove duplicate rows
    df_clean = df.drop_duplicates().copy()

    # Handle missing values:
    # Numeric columns -> median
    # Categorical columns -> mode
    numeric_cols = df_clean.select_dtypes(include=np.number).columns
    categorical_cols = df_clean.select_dtypes(exclude=np.number).columns

    for col in numeric_cols:
        if df_clean[col].isnull().any():
            df_clean[col] = df_clean[col].fillna(df_clean[col].median())

    for col in categorical_cols:
        if df_clean[col].isnull().any() and not df_clean[col].mode().empty:
            df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

    print("Original shape:", df.shape)
    print("Cleaned shape:", df_clean.shape)
    print("Cleaning completed.")
```

## 5. Numerical Analysis

```python
if 'df_clean' in globals():
    numeric_cols = df_clean.select_dtypes(include=np.number).columns

    if len(numeric_cols) > 0:
        display(df_clean[numeric_cols].describe().T)
    else:
        print("No numerical columns found.")
```

## 6. Categorical Analysis

```python
if 'df_clean' in globals():
    categorical_cols = df_clean.select_dtypes(exclude=np.number).columns

    for col in categorical_cols[:5]:
        print(f"\nValue counts for: {col}")
        display(df_clean[col].value_counts().head(10))
```

## 7. Visualizations

The following charts are generated automatically from available numerical and categorical columns.


```python
if 'df_clean' in globals():
    numeric_cols = df_clean.select_dtypes(include=np.number).columns

    if len(numeric_cols) > 0:
        col = numeric_cols[0]
        plt.figure(figsize=(8, 5))
        plt.hist(df_clean[col].dropna(), bins=20)
        plt.title(f"Distribution of {col}")
        plt.xlabel(col)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()
    else:
        print("No numerical column available for histogram.")
```

```python
if 'df_clean' in globals():
    categorical_cols = df_clean.select_dtypes(exclude=np.number).columns

    if len(categorical_cols) > 0:
        col = categorical_cols[0]
        counts = df_clean[col].value_counts().head(10)

        plt.figure(figsize=(9, 5))
        counts.plot(kind="bar")
        plt.title(f"Top Categories in {col}")
        plt.xlabel(col)
        plt.ylabel("Count")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()
    else:
        print("No categorical column available for bar chart.")
```

```python
if 'df_clean' in globals():
    numeric_cols = df_clean.select_dtypes(include=np.number).columns

    if len(numeric_cols) >= 2:
        plt.figure(figsize=(8, 5))
        plt.scatter(df_clean[numeric_cols[0]], df_clean[numeric_cols[1]], alpha=0.6)
        plt.title(f"{numeric_cols[0]} vs {numeric_cols[1]}")
        plt.xlabel(numeric_cols[0])
        plt.ylabel(numeric_cols[1])
        plt.tight_layout()
        plt.show()
    else:
        print("At least two numerical columns are required for a scatter plot.")
```

## 8. Correlation Analysis

```python
if 'df_clean' in globals():
    numeric_cols = df_clean.select_dtypes(include=np.number).columns

    if len(numeric_cols) >= 2:
        correlation = df_clean[numeric_cols].corr()
        display(correlation)

        plt.figure(figsize=(8, 6))
        plt.imshow(correlation, interpolation="nearest")
        plt.colorbar()
        plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=45, ha="right")
        plt.yticks(range(len(correlation.columns)), correlation.columns)
        plt.title("Correlation Matrix")
        plt.tight_layout()
        plt.show()
    else:
        print("At least two numerical columns are required for correlation analysis.")
```

## 9. Key Insights

```python
if 'df_clean' in globals():
    print("KEY DATASET INSIGHTS")
    print("=" * 40)
    print(f"Total records: {len(df_clean)}")
    print(f"Total columns: {len(df_clean.columns)}")

    numeric_cols = df_clean.select_dtypes(include=np.number).columns

    for col in numeric_cols[:5]:
        print(f"- {col}:")
        print(f"  Mean   = {df_clean[col].mean():.2f}")
        print(f"  Median = {df_clean[col].median():.2f}")
        print(f"  Min    = {df_clean[col].min():.2f}")
        print(f"  Max    = {df_clean[col].max():.2f}")

    categorical_cols = df_clean.select_dtypes(exclude=np.number).columns
    for col in categorical_cols[:3]:
        if not df_clean[col].mode().empty:
            print(f"- Most frequent {col}: {df_clean[col].mode()[0]}")
```

## 10. Save Cleaned Dataset

The cleaned dataset is saved as `cleaned_data.csv`.


```python
if 'df_clean' in globals():
    OUTPUT_FILE = "cleaned_data.csv"
    df_clean.to_csv(OUTPUT_FILE, index=False)
    print(f"Cleaned dataset saved as: {OUTPUT_FILE}")
```
